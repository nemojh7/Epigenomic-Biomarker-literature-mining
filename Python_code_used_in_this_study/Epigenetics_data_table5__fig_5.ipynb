```python
import matplotlib.pyplot as plt

# ==============================
# 데이터 입력
# ==============================

data = [
    ("5C", 4476, 2474),
    ("ChIA-PET", 2737, 477),
    ("Hi-C", 14612, 1026),
    ("DNase-seq", 3079, 689),
    ("FAIRE-seq", 1523, 433),
    ("ATAC-seq", 13630, 1174),
    ("ChIP-seq", 17757, 1840),
    ("WGBS", 3964, 555),
    ("RRBS", 3593, 490),
    ("methylation array", 1052, 505),
    ("Computationall", 9651, 2511),
    ("Predictions", 5897, 1523),
    ("RNA-seq", 23115, 3063),
    ("CLIP-seq", 1115, 483),
    ("RIP-seq", 1854, 423),
    ("Epigenomics", 2692, 1285),
]

methods = [d[0] for d in data]
occurrences = [d[1] for d in data]
files = [d[2] for d in data]

# ==============================
# 그래프 생성
# ==============================

plt.figure(figsize=(14, 7))

ax1 = plt.gca()

# Occurrences (파란색)
ax1.plot(methods, occurrences, marker='o', linewidth=2)
ax1.set_ylabel("Number of Occurrences")
ax1.tick_params(axis='y')

# 수치 표시 (Occurrences)
for i, y in enumerate(occurrences):
    ax1.text(i, y, str(y), ha='center', va='bottom', fontsize=8)

# 두 번째 y축 생성
ax2 = ax1.twinx()

# Files (빨간색)
ax2.plot(methods, files, marker='s', linewidth=2)
ax2.set_ylabel("Number of Files")
ax2.tick_params(axis='y')

# 수치 표시 (Files)
for i, y in enumerate(files):
    ax2.text(i, y, str(y), ha='center', va='bottom', fontsize=8)

plt.xticks(rotation=45, ha="right")
plt.title("Epigenomic Experimental Method Keywords\nOccurrences vs Number of Files")

plt.tight_layout()

# ==============================
# 저장 (논문용 고해상도)
# ==============================

plt.savefig("Figure_Table5_DualAxis.png", dpi=600)
plt.savefig("Figure_Table5_DualAxis.pdf")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==============================
# 데이터 입력
# ==============================

data = [
    ("5C", 4476, 2474),
    ("ChIA-PET", 2737, 477),
    ("Hi-C", 14612, 1026),
    ("DNase-seq", 3079, 689),
    ("FAIRE-seq", 1523, 433),
    ("ATAC-seq", 13630, 1174),
    ("ChIP-seq", 17757, 1840),
    ("WGBS", 3964, 555),
    ("RRBS", 3593, 490),
    ("methylation array", 1052, 505),
    ("Computationall", 9651, 2511),
    ("Predictions", 5897, 1523),
    ("RNA-seq", 23115, 3063),
    ("CLIP-seq", 1115, 483),
    ("RIP-seq", 1854, 423),
    ("Epigenomics", 2692, 1285),
]

methods = [d[0] for d in data]
occurrences = [d[1] for d in data]
files = [d[2] for d in data]

x = np.arange(len(methods))

# ==============================
# 그래프 생성
# ==============================

plt.figure(figsize=(15, 8))

ax1 = plt.gca()

# 색상 설정 (논문용 세련된 색)
color_occ = "#1f4e79"   # 진한 블루
color_files = "#c00000" # 진한 레드

# Occurrences
ax1.plot(x, occurrences, marker='o', markersize=6,
         linewidth=2.5, color=color_occ, label="Occurrences")

ax1.set_ylabel("Number of Occurrences", fontsize=12, color=color_occ)
ax1.tick_params(axis='y', labelcolor=color_occ)

# Occurrences 수치 표시
for i, y in enumerate(occurrences):
    ax1.text(i, y + max(occurrences)*0.01, f"{y}",
             ha='center', va='bottom', fontsize=8, color=color_occ)

# 두 번째 y축
ax2 = ax1.twinx()

# Files
ax2.plot(x, files, marker='s', markersize=6,
         linewidth=2.5, color=color_files, label="Number of Files")

ax2.set_ylabel("Number of Files", fontsize=12, color=color_files)
ax2.tick_params(axis='y', labelcolor=color_files)

# Files 수치 표시
for i, y in enumerate(files):
    ax2.text(i, y + max(files)*0.01, f"{y}",
             ha='center', va='bottom', fontsize=8, color=color_files)

# X축 설정
plt.xticks(x, methods, rotation=45, ha="right")

plt.title("Occurrence Frequency and Corresponding Document Counts\nof Epigenomic Experimental Method Keywords",
          fontsize=14)

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()

# ==============================
# 저장
# ==============================

plt.savefig("Figure_Table5_DualAxis_Improved.png", dpi=600)
plt.savefig("Figure_Table5_DualAxis_Improved.pdf")

plt.show()
```
