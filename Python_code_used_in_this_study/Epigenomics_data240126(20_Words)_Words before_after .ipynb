# genetic loci_pattern, file_name, loci_pattern(before/after) word 추출

```python
import os
import re

def extract_context(text, loci, window_size=20):
  # 입력된 텍스트를 공백을 기준으로 분할하여 단어 목록 생성
    words = text.split()
    # enumerate : 인자로 넘어온 목록을 기준으로 인덱스와 원소를 차례대로 접근하게 해주는 반복자(iterator) 객체를 반환해주는 함수
    # 단어 목록을 순회하면서, 각 단어와 그 위치 i를 가지고 오기
    for i, word in enumerate(words):
        if loci in word:
          # 현재 단어의 위치에서 window_size만큼 앞으로 간 위치를 계산, 이 값이 0보다 작지 않도록 하기
            start = max(0, i - window_size)
            # 현재 단어의 위치에서 window_size만큼 뒤로 간 위치를 계산하되, 이 값이 전체 단어 수를 초과하지 않도록 하기
            end = min(len(words), i + window_size + 1)
            return ' '.join(words[start:i]), ' '.join(words[i+1:end])
    return "", ""

# 데이터 파일 경로
text_folder_path = 'C:/Users/jhang/FullText_Epigenomics_data240126'

# loci_pattern의 패턴 지정
loci_pattern = re.compile(r'\d+[pq]\d+(\.\d+)?')

# 찾은 모든 위치, 파일 이름, 제목 및 컨텍스트 저장
all_data = []

for file_name in os.listdir(text_folder_path):
    # endswith() : 메서드 사용하여 어떤 문자열에서 특정 문자열로 끝나는지 확인 후, 그 결과 true 혹은 false 로 반환
    if file_name.endswith('.txt'):
        file_path = os.path.join(text_folder_path, file_name)
        with open(file_path, 'r', encoding='utf-8') as file:
            text = file.read()

            for match in loci_pattern.finditer(text):
                loci = match.group()
                before, after = extract_context(text, loci)
                all_data.append((file_name, loci, before, after))

# 찾은 file_name, loci, before, after 출력
for file_name, loci, before, after in all_data:
    print(f"File: {file_name}, Loci: {loci}")
    print(f"Words before: {before}")
    print(f"Words after: {after}\n")
```
